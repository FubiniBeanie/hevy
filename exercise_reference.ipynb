The purpose of this file is to get a large data set of information from the Hevy templates. This data set will be used in the file workout_scores to then generate a function that creates the "muscle score". This file will need to have access to the Hevy API. Also we will need to make an additional part of this file that will update the database in the case that we create custom exercises. 

```python
import getpass
import json
import urllib.request
from pathlib import Path

project_folder = Path("C:/Users/AndyM/Projects/hevy")
data_folder = project_folder / "data"
data_folder.mkdir(parents=True, exist_ok=True)

api_key = getpass.getpass("Hevy API key: ")

templates = {}
page = 1

while True:
    url = (
        "https://api.hevyapp.com/v1/exercise_templates"
        f"?page={page}&pageSize=100"
    )

    request = urllib.request.Request(
        url,
        headers={"api-key": api_key},
    )

    with urllib.request.urlopen(request, timeout=30) as response:
        data = json.load(response)

    # Add this page's templates to our dictionary.
    for template in data["exercise_templates"]:
        template_id = template["id"]
        templates[template_id] = template

    print(
        f"Downloaded page {page} of {data['page_count']} "
        f"— {len(templates)} templates collected"
    )

    # Stop after the last page.
    if page >= data["page_count"]:
        break

    page += 1

# Save only after all pages have been downloaded successfully.
output_path = data_folder / "exercise_templates.json"

with output_path.open("w", encoding="utf-8") as file:
    json.dump(templates, file, indent=2, ensure_ascii=False)

print(f"Saved {len(templates)} templates to: {output_path}")
```
